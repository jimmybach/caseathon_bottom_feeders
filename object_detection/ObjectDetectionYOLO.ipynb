```python
from ultralytics import YOLO
import os

#os.chdir('drive/Shareddrives/CaseathonFall2025/ObjectDetectionYOLO/object_detection_dataset')
# Load a model
model = YOLO("yolo11x.pt")  # load a pretrained model (recommended for training)
os.chdir('drive/Shareddrives/CaseathonFall2025/ObjectDetectionYOLO/object_detection_dataset')

# Train the model
results = model.train(data="data.yaml", epochs=100, imgsz=640)
```

```python
model.val()
```

```python
results = model.predict(
    source="test/images",  # folder or glob pattern
    save=True,                             # save predicted images
    save_txt=True,                         # save bounding boxes to .txt
    conf=0.25,                             # confidence threshold
    iou=0.95,                              # NMS IoU threshold
    imgsz=640,                             # inference image size
    device=0                               # or 'cpu'
)
```

```python
model.tune(epochs=20,resume=True)
```
